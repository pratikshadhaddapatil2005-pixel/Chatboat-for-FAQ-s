```python
from ultralytics import YOLO
from google.colab import files
import cv2
import os
```

```python
uploaded = files.upload()

video_path = list(uploaded.keys())[0]

print("Video uploaded successfully!")
print("Input video:", video_path)
```

```python
model = YOLO("yolov8n.pt")

print("YOLO model loaded successfully!")
```

```python
import cv2
import os

# Open input video
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    raise Exception("Could not open the input video.")

# Get video information
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

if fps <= 0:
    fps = 30

total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

print("Video width:", width)
print("Video height:", height)
print("FPS:", fps)
print("Total frames:", total_frames)

# Output file
output_video = "/content/Task4_Object_Detection_Tracking.mp4"

# Create video writer
fourcc = cv2.VideoWriter_fourcc(*"mp4v")
out = cv2.VideoWriter(
    output_video,
    fourcc,
    fps,
    (width, height)
)

if not out.isOpened():
    raise Exception("Could not create output video.")

frame_number = 0

while True:

    success, frame = cap.read()

    if not success:
        break

    frame_number += 1

    # YOLO object detection + ByteTrack tracking
    results = model.track(
        frame,
        persist=True,
        tracker="bytetrack.yaml",
        verbose=False
    )

    # Draw bounding boxes, labels and tracking IDs
    annotated_frame = results[0].plot()

    # Save processed frame
    out.write(annotated_frame)

    # Show progress
    if frame_number % 50 == 0 or frame_number == total_frames:
        print(f"Processed {frame_number}/{total_frames} frames")

# Release resources
cap.release()
out.release()

print("\n✅ Object detection and tracking completed!")
print("Output video:", output_video)
print("Output file exists:", os.path.exists(output_video))
print(
    "Output size:",
    round(os.path.getsize(output_video) / (1024 * 1024), 2),
    "MB"
)
```

```python
import cv2

input_video = "task4_input.mp4"
output_20sec = "/content/task4_20sec.mp4"

cap = cv2.VideoCapture(input_video)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

max_frames = int(fps * 20)

fourcc = cv2.VideoWriter_fourcc(*"mp4v")
out = cv2.VideoWriter(
    output_20sec,
    fourcc,
    fps,
    (width, height)
)

count = 0

while count < max_frames:
    success, frame = cap.read()

    if not success:
        break

    out.write(frame)
    count += 1

cap.release()
out.release()

print("✅ 20-second video created!")
print("Frames:", count)
print("FPS:", fps)
print("Duration:", round(count / fps, 2), "seconds")
print("File:", output_20sec)
```

```python
from ultralytics import YOLO
import cv2
import os

# Use the 20-second video
input_video = "/content/task4_20sec.mp4"

# Load YOLO model
model = YOLO("yolov8n.pt")

# Open video
cap = cv2.VideoCapture(input_video)

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

output_video = "/content/Task4_Final_Object_Detection_Tracking.mp4"

fourcc = cv2.VideoWriter_fourcc(*"mp4v")
out = cv2.VideoWriter(
    output_video,
    fourcc,
    fps,
    (width, height)
)

frame_count = 0

while True:
    success, frame = cap.read()

    if not success:
        break

    # Object detection + tracking
    results = model.track(
        frame,
        persist=True,
        tracker="bytetrack.yaml",
        verbose=False
    )

    # Draw boxes, labels and tracking IDs
    annotated_frame = results[0].plot()

    out.write(annotated_frame)

    frame_count += 1

cap.release()
out.release()

print("✅ Task 4 completed successfully!")
print("Frames processed:", frame_count)
print("Output video:", output_video)
print("File exists:", os.path.exists(output_video))
print("File size:", round(os.path.getsize(output_video) / (1024*1024), 2), "MB")
```
