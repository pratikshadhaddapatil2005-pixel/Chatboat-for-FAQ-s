```python

!pip install -q gradio deep-translator gTTS

import gradio as gr
from deep_translator import GoogleTranslator
from gtts import gTTS
import tempfile

# Supported languages
LANGUAGES = {
    "English": "en",
    "Hindi": "hi",
    "Kannada": "kn",
    "Marathi": "mr",
    "Tamil": "ta",
    "Telugu": "te",
    "Malayalam": "ml",
    "Bengali": "bn",
    "Gujarati": "gu",
    "Punjabi": "pa",
    "Urdu": "ur",
    "French": "fr",
    "German": "de",
    "Spanish": "es",
    "Italian": "it",
    "Japanese": "ja",
    "Korean": "ko",
    "Chinese": "zh-CN",
    "Arabic": "ar",
    "Russian": "ru"
}


# Translation function
def translate_text(text, source, target):

    if not text.strip():
        return "Please enter some text."

    try:
        translator = GoogleTranslator(
            source=LANGUAGES[source],
            target=LANGUAGES[target]
        )

        result = translator.translate(text)

        return result

    except Exception as e:
        return "Error: " + str(e)


# Text-to-speech function
def text_to_speech(text, language):

    if not text.strip():
        return None

    try:
        audio_file = tempfile.NamedTemporaryFile(
            delete=False,
            suffix=".mp3"
        )
        audio_file.close()

        speech = gTTS(
            text=text,
            lang=LANGUAGES[language]
        )

        speech.save(audio_file.name)

        return audio_file.name

    except Exception:
        return None


# User Interface
with gr.Blocks(title="AI Language Translation Tool") as app:

    gr.Markdown("""
    # 🌐 AI Language Translation Tool

    Translate text between multiple languages using AI.
    """)

    with gr.Row():

        source = gr.Dropdown(
            choices=list(LANGUAGES.keys()),
            value="English",
            label="Source Language"
        )

        target = gr.Dropdown(
            choices=list(LANGUAGES.keys()),
            value="Kannada",
            label="Target Language"
        )

    input_text = gr.Textbox(
        label="Enter Text",
        placeholder="Type your text here...",
        lines=5
    )

    translate_btn = gr.Button(
        "🔄 Translate",
        variant="primary"
    )

    output_text = gr.Textbox(
        label="Translated Text",
        lines=5
    )

    translate_btn.click(
        fn=translate_text,
        inputs=[input_text, source, target],
        outputs=output_text
    )

    speak_btn = gr.Button("🔊 Text to Speech")

    audio = gr.Audio(
        label="Speech Output",
        type="filepath"
    )

    speak_btn.click(
        fn=text_to_speech,
        inputs=[output_text, target],
        outputs=audio
    )

    gr.Markdown("""
    ### Features
    ✅ Multiple language translation
    ✅ Google Translation service
    ✅ Text-to-speech
    ✅ Simple user interface
    """)


# Start application
app.launch(share=True)
```
