```python
# ==========================================
# CODEALPHA - TASK 2
# CHATBOT FOR FAQs
# ==========================================

import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# ------------------------------------------
# 1. FAQ DATA
# ------------------------------------------

faq_questions = [
    "What is artificial intelligence?",
    "What is machine learning?",
    "What is Python?",
    "What is NLP?",
    "What is a chatbot?",
    "What is deep learning?",
    "What is data science?",
    "What is computer vision?",
    "What are the applications of AI?",
    "How can I learn artificial intelligence?"
]

faq_answers = [
    "Artificial Intelligence is a technology that enables computers to perform tasks that normally require human intelligence.",

    "Machine Learning is a branch of AI that allows computers to learn from data and make predictions.",

    "Python is a popular programming language used for AI, machine learning, data science and web development.",

    "NLP stands for Natural Language Processing. It helps computers understand human language.",

    "A chatbot is a computer program that communicates with users and answers their questions.",

    "Deep Learning is a type of machine learning that uses neural networks with multiple layers.",

    "Data Science is the process of collecting, analyzing and interpreting data to find useful information.",

    "Computer Vision is an AI technology that helps computers understand images and videos.",

    "AI is used in healthcare, education, banking, robotics, recommendation systems and many other areas.",

    "You can learn AI by studying Python, mathematics, machine learning and deep learning and by creating projects."
]


# ------------------------------------------
# 2. TEXT PREPROCESSING
# ------------------------------------------

def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    return text


clean_questions = []

for question in faq_questions:
    clean_questions.append(clean_text(question))


# ------------------------------------------
# 3. TF-IDF MODEL
# ------------------------------------------

vectorizer = TfidfVectorizer()

faq_vectors = vectorizer.fit_transform(clean_questions)


# ------------------------------------------
# 4. CHATBOT FUNCTION
# ------------------------------------------

def get_answer(user_question):

    user_question = clean_text(user_question)

    user_vector = vectorizer.transform([user_question])

    similarity = cosine_similarity(
        user_vector,
        faq_vectors
    )

    best_match = similarity.argmax()

    score = similarity[0][best_match]

    if score < 0.10:
        return "Sorry, I don't know the answer to that question."

    return faq_answers[best_match]


# ------------------------------------------
# 5. TEST QUESTIONS
# ------------------------------------------

print("====================================")
print("       AI FAQ CHATBOT")
print("====================================")

print("\nQuestion: What is AI?")
print("Answer:", get_answer("What is AI?"))

print("\nQuestion: What is Python?")
print("Answer:", get_answer("Tell me about Python"))

print("\nQuestion: What is NLP?")
print("Answer:", get_answer("Explain NLP"))


# ------------------------------------------
# 6. SIMPLE CHATBOT
# ------------------------------------------

print("\n====================================")
print("       CHATBOT IS READY")
print("====================================")
print("Type 'exit' to stop the chatbot.\n")


while True:

    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Bot: Thank you! Goodbye.")
        break

    answer = get_answer(user_input)

    print("Bot:", answer)
```
